```python
import numpy as np
from ase import Atoms, io
from DynamicTopology.ase import DynamicTopology
from DynamicTopology.core import ReactionSet
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
reaction_set = ReactionSet("../datasets/HCombustion/HCombustion.json")
reactions = sorted(Path("../datasets/HCombustion/reactions/").glob("*.xyz"))
reactions
```

```python
def compute_energies(frames: list[Atoms]):
    energies = []
    for frame in frames:
        frame.calc = DynamicTopology(frame, reaction_set)
        energy = frame.get_potential_energy()
        energies.append(energy)
    return energies
```

```python
for reaction in reactions:
    frames = io.read(reaction, index=slice(None))
    R, T, P = frames[0], frames[1], frames[-1]

    R_interp = []
    P_interp = []

    for t in np.linspace(0, 1, 10):
        frame = R.copy()
        frame.positions = (1 - t) * R.positions + t * T.positions
        frame.set_cell([50, 50, 50])
        R_interp.append(frame)

        frame = P.copy()
        frame.positions = t * P.positions + (1 - t) * T.positions
        frame.set_cell([50, 50, 50])
        P_interp.append(frame)

    for t in np.linspace(0, 1, 10):
        frame = R.copy()
        frame.positions = t * P.positions + (1 - t) * T.positions
        frame.set_cell([50, 50, 50])
        R_interp.append(frame)

        frame = P.copy()
        frame.positions = (1 - t) * R.positions + t * T.positions
        frame.set_cell([50, 50, 50])
        P_interp.insert(0, frame)

    print(R_interp)
    ER_interp = compute_energies(R_interp)
    EP_interp = compute_energies(P_interp)

    plt.figure()
    plt.scatter(
        range(len(ER_interp)),
        ER_interp,
    )
    plt.scatter(
        range(len(EP_interp)),
        EP_interp,
    )
    plt.title(reaction.name)
    plt.show()
```
